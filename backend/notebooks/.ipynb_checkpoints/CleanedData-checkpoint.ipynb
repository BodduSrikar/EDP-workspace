```python
import pandas as pd
import os

raw_folder = r'C:\EDP-workspace\datasetFornews'
project_folder = r'C:\EDP-workspace\Fake-News-Detector\data'

os.makedirs(project_folder, exist_ok=True)

fake_df = pd.read_csv(os.path.join(raw_folder, 'Fake.csv'))
true_df = pd.read_csv(os.path.join(raw_folder, 'True.csv'))

fake_df['label'] = 1
true_df['label'] = 0

df = pd.concat([fake_df, true_df], ignore_index=True)
df = df.dropna().drop_duplicates()

output_path = os.path.join(project_folder, 'cleaned_news_dataset.csv')
df.to_csv(output_path, index=False)

print("SUCCESS!")
print(f"Cleaned dataset saved at: {output_path}")
```
